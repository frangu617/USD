## Problem 3.8.

Construct the sampling distribution of the sample proportion of heads, for flipping a balanced
coin (a) once; (b) twice; (c) three times; (d) four times. Describe how the shape changes
as the number of flips $n$ increases. What would happen if $n$ kept growing? Why?


Let $H$ be the number of heads and $\hat p=H/n$ be the sample proportion of heads. For a balanced coin, $p=0.50$, and independent flips give:

$$
H\sim\operatorname{Binomial}(n,0.50),\qquad
P\left(\hat p=\frac{h}{n}\right)=\binom{n}{h}(0.50)^n
$$

Here, $\binom{n}{h}$ counts the different ways to get $h$ heads in $n$ flips. The exact distributions are:

| Part | Number of flips $n$ | Possible proportions of heads | Corresponding probabilities |
| --- | ---: | --- | --- |
| (a) | 1 | $0,1$ | $1/2,1/2$ |
| (b) | 2 | $0,1/2,1$ | $1/4,2/4,1/4$ |
| (c) | 3 | $0,1/3,2/3,1$ | $1/8,3/8,3/8,1/8$ |
| (d) | 4 | $0,1/4,1/2,3/4,1$ | $1/16,4/16,6/16,4/16,1/16$ |

For example, with four flips there are $2^4=16$ equally likely sequences. Six have exactly two heads, so $P(\hat p=0.50)=6/16=37.50\%$. Fractions in the table keep the probabilities exact.

![Exact sampling distributions for one through four coin flips](charts/problem3_8.png)

The distributions are symmetric around 0.50. With more flips, the proportions become more concentrated near 0.50 and the shape becomes more bell-shaped. The center and standard error are:

$$
E(\hat p)=0.50,\qquad SE(\hat p)=\sqrt{\frac{0.50(1-0.50)}{n}}=\frac{0.50}{\sqrt n}
$$

For $n=1,2,3,4$, the standard errors are 0.50, 0.35, 0.29, and 0.25. If $n$ kept growing, the proportion would tend to get closer to 0.50 as its standard error approaches zero. This agrees with the law of large numbers; the Central Limit Theorem explains the approximately normal shape of the standardized proportions for large $n$.

## Problem 3.14.

On each bet in a sequence of bets, you win 1 dollar with probability 0.50 and lose 1 dollar (i.e., win negative 1 dollar) with probability 0.50. Let $Y$ denote the total of your winnings and losings after 100 bets. Giving your reasoning, state the approximate distribution of $Y.$ 

Let $X_i$ be the winnings on bet $i$. Each bet gives $+1$ dollar or $-1$ dollar with equal probability. Assuming the bets are independent, the mean and variance for one bet are:

$$
E(X_i)=1(0.50)+(-1)(0.50)=0
$$

$$
\operatorname{Var}(X_i)=E(X_i^2)-[E(X_i)]^2
=1^2(0.50)+(-1)^2(0.50)-0^2=1
$$

The total after 100 bets is $Y=\sum_{i=1}^{100}X_i$. Means and variances add for independent bets:

$$
E(Y)=100(0)=0,\qquad\operatorname{Var}(Y)=100(1)=100
$$

$$
SD(Y)=\sqrt{100}=10.00\text{ dollars}
$$

By the Central Limit Theorem, the total is approximately normal:

$$
Y\approx N(0,10^2)
$$

So the approximate distribution has mean 0.00 dollars and standard deviation 10.00 dollars. The exact total is discrete: if $W\sim\operatorname{Binomial}(100,0.50)$ is the number of wins, then $Y=2W-100$, so only even-dollar totals from $-100$ to $100$ are possible. The normal distribution is an approximation to these totals.

Small samples have more sampling variability than large samples. For independent observations, the standard errors of a sample mean and a sample proportion are:

$$

$$

Both get larger when $n$ gets smaller. For example, a sample one-fourth as large has twice the standard error, when the population variability is the same.

This means a few additional cases can change a cancer rate a lot in a small population, and a few students can change the average improvement a lot in a small school.
